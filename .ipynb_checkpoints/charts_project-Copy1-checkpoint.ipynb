# **Технический файл**

```python
import os
import pandas as pd
import requests
from dotenv import load_dotenv

# грузим переменные окружения из .env
load_dotenv()

API_URL = os.getenv("API_URL")
DATE_BEGIN = os.getenv("DATE_BEGIN")
DATE_END = os.getenv("DATE_END")
```

### **3. Запросы к API.**

```python
# визиты
visits_resp = requests.get(f"{API_URL}/visits", params={"begin": DATE_BEGIN, "end": DATE_END})
visits = pd.DataFrame(visits_resp.json())

# регистрации
regs_resp = requests.get(f"{API_URL}/registrations", params={"begin": DATE_BEGIN, "end": DATE_END})
regs = pd.DataFrame(regs_resp.json())
```

### **4. Расчет метрик.**

```python
# чистим ботов и дубли визитов
visits_clean = visits[~visits["user_agent"].str.contains("bot", case=False, na=False)].copy()
visits_clean = visits_clean.sort_values("datetime").drop_duplicates(subset="visit_id", keep="last")
visits_clean["date_group"] = pd.to_datetime(visits_clean["datetime"]).dt.floor("D")

# print(visits_clean.shape)
```

```python
# группировка по дате и платформе
regs["date_group"] = pd.to_datetime(regs["datetime"]).dt.floor("D")

visits_agg = visits_clean.groupby(["date_group", "platform"]).size().reset_index(name="visits")
regs_agg = regs.groupby(["date_group", "platform"]).size().reset_index(name="registrations")
```

```python
# считаем конверсию
conversion_df = visits_agg.merge(regs_agg, on=["date_group", "platform"], how="left")
conversion_df["registrations"] = conversion_df["registrations"].fillna(0).astype(int)
conversion_df["conversion"] = conversion_df["registrations"] / conversion_df["visits"] * 100
conversion_df = conversion_df.sort_values("date_group").reset_index(drop=True)
```

### **5. Добавляем рекламы.**

```python
# суммирую визиты и регистрации по датам
daily_totals = conversion_df.groupby("date_group")[["visits", "registrations"]].sum().reset_index()
```

```python
# реклама
ads = pd.read_csv("./ads.csv")
ads["date_group"] = pd.to_datetime(ads["date"]).dt.floor("D")

# приклеиваем рекламу
ads_df = daily_totals.merge(ads[["date_group", "utm_campaign", "cost"]], on="date_group", how="left")
```

```python
# без рекламы
ads_df["cost"] = ads_df["cost"].fillna(0).astype(int)
ads_df["utm_campaign"] = ads_df["utm_campaign"].fillna("none")
```

```python
# сортировка
ads_df = ads_df.sort_values("date_group").reset_index(drop=True)
ads_df = ads_df[["date_group", "visits", "registrations", "cost", "utm_campaign"]]
```

```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(figsize=(20, 8))
ax.plot(ads_df["date_group"], ads_df["visits"], color="black", marker="o")

campaigns = [c for c in ads_df["utm_campaign"].unique() if c != "none"]
colors = plt.cm.tab10.colors
for i, campaign in enumerate(campaigns):
    dates = ads_df.loc[ads_df["utm_campaign"] == campaign, "date_group"]
    ax.axvspan(dates.min(), dates.max(), color=colors[i], alpha=0.3)

active = ads_df[ads_df["utm_campaign"] != "none"]
gaps = ads_df[ads_df["utm_campaign"] == "none"]
avg_campaign = active["visits"].mean()
avg_gap = gaps["visits"].mean()

ax.axhline(avg_campaign, color="#2F80ED", linestyle=(0, (0.5, 1.75)), linewidth=6, dash_capstyle='round')
ax.axhline(avg_gap, color="#70777D", linestyle=(0, (0.5, 1.5)), linewidth=7, dash_capstyle='round')

ax.set_xlabel("Date")
ax.set_ylabel("Unique Visits")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.savefig("./charts/visits_with_campaigns_v2.png")
plt.show()
```

```python
conv_pivot = conversion_df.pivot(index="date_group", columns="platform", values="conversion").sort_index()

# скользящее среднее за 7 дней, чтобы видеть устойчивый тренд, а не шум одного дня
rolling = conv_pivot[["android", "ios"]].rolling(7).mean()

# ищем растянутый период, где обе платформы устойчиво ниже 70%
low_period = rolling[(rolling["android"] < 70) & (rolling["ios"] < 70)]
print("Найдено дней в устойчивой просадке:", len(low_period))
print("Период:", low_period.index.min().date(), "—", low_period.index.max().date())
```

```python
# визиты и реклама в этот же период — были ли они в норме, раз конверсия упала именно из-за регистрации, а не оттока трафика
period_mask = (ads_df["date_group"] >= "2023-06-27") & (ads_df["date_group"] <= "2023-07-28")
period_data = ads_df[period_mask]

print("Средние визиты в период просадки:", round(period_data["visits"].mean()))
print("Средние визиты за весь проект:", round(ads_df["visits"].mean()))
print("Была ли активна реклама весь период:", (period_data["utm_campaign"] != "none").all())
print("Кампания(и) в этот период:", period_data["utm_campaign"].unique())

# для контраста - средняя конверсия android/ios ДО и ПОСЛЕ этого периода
before_after = conv_pivot[(conv_pivot.index < "2023-06-27") | (conv_pivot.index > "2023-07-28")]
print("Средняя конверсия android вне периода просадки:", round(before_after["android"].mean(), 1))
print("Средняя конверсия android в период просадки:", round(rolling.loc["2023-06-27":"2023-07-28", "android"].mean(), 1))
```

```python
print("Средняя конверсия ios вне периода просадки:", round(before_after["ios"].mean(), 1))
print("Средняя конверсия ios в период просадки:", round(rolling.loc["2023-06-27":"2023-07-28", "ios"].mean(), 1))
```

```python
web_before_after = conv_pivot[(conv_pivot.index < "2023-06-27") | (conv_pivot.index > "2023-07-28")]["web"].mean()
web_in_period = rolling.index  # rolling не считался для web, посчитаем отдельно
web_rolling_period = conv_pivot.loc["2023-06-27":"2023-07-28", "web"].mean()
print("Web вне периода:", round(web_before_after, 1))
print("Web в период просадки:", round(web_rolling_period, 1))
```

```python
import seaborn as sns
sns.set_style("whitegrid")

conv_pivot = conversion_df.pivot(index="date_group", columns="platform", values="conversion")

fig, axes = plt.subplots(3, 1, figsize=(20, 14), sharex=True)

period_start = "2023-06-27"
period_end = "2023-07-28"

for ax, platform in zip(axes, ["android", "ios", "web"]):
    ax.plot(conv_pivot.index, conv_pivot[platform], marker="o", label=platform)
    ax.axvspan(period_start, period_end, color="red", alpha=0.15)

    for x, y in zip(conv_pivot.index, conv_pivot[platform]):
        ax.annotate(f"{y:.0f}%", (x, y), textcoords="offset points", xytext=(0, 5), ha="center", fontsize=7)

    ax.set_title(f"Conversion {platform}")
    ax.set_ylabel("Conversion (%)")
    ax.legend()

axes[-1].set_xlabel("Date")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.savefig("./charts/conversion_by_platform_v2.png")
plt.show()
```
